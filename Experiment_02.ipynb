## Loops-for,range(), while

```python
for i in range(1,10):
  print(i)
```

```python
for i in range(0,30,3):
  print(i)
```

```python
for i in range(10,0,-2):
  print(i)
```

```python
i = 0

while i < 20:
  print(i)
  i +=4
```

```python
#The loop will evaluated 10 time 0 to 9
i = 0
while i<10:
  print(i)
  i +=1
```

## Conditions-if/elif/else,modulo(%)

```python
i = 1
while i<20:
  if i%2 == 0:
    print(i)

  i +=1
```

```python
for i in range(1,15):
  if i%3 == 0:
    print("Fizz")
  elif i%5 == 0:
    print("FizzBuzz")
  elif i%15 == 0:
     print(i)
```

```python
for i in range(0,5):
  if i%3 ==0:
    print(i)
  elif i%3 == 1:
    print(i+10)
  else:
    print(i-10)
```
